## Exercise 1:

Write a program that reads an email address from the keyboard with the format `user@domain`. After that, it must print the username in one line and the domain in the next line.

### Expected Input:
```text
user@domain.com
```

### Expected Output:
```text
user
domain.com
```

---

````python
def read_email():
    go_on = True
    while go_on:
        x = input("Type an email address: ")
        if x.count('@') != 1:
            print('Incorrect format')
        else:
            name, domain = x.split('@')
            print (f"The name is {name} \nThe domain is {domain}")
            go_on = False

read_email()
````

## Excercise 2:

Write a program that reads from keyboard the following data:
* The name of a product.

* The unit price of the product (it could contain decimals).

* The number of units of the product (it must be an integer number).

Then it must print on the console the following text:

`<product_name>: <number_units> units x <unit_price>€ = <total>€`
where `<number_units>` is printed with five digits, `<unit_price>` is printed with 6 digits
for the integer part and 2 digits for the decimal part and `<total>` is printed with 8 digits
for the integer part and 2 digits for the decimal part.

````python
def read_item():
    name = input("Type Product name: ")

    go_on = True
    while go_on:
        try:
            price = input("Type Unit price: ")
            price = float(price)
            go_on = False
        except ValueError:
            print("Incorrect format")

    go_on = True
    while go_on:
        try:
            n_units = input("Type Number of Units: ")
            n_units = int(n_units)
            go_on = False
        except ValueError:
            print('Incorrect format')

    total = price * n_units
    print(f"{name}: {n_units} x {price:6.2f} € = {total:8.2f}")

read_item()
````

## Excercise 3:

Write a program to print all the prime numbers between `a and b`, where a and b are
**integers** read from the keyboard.

````python
def primes(a,b):
    prms = [2]
    for i in range(3, b+1):
        prmp = True
        for p in prms:
            if i % p == 0:
                prmp = False
                break
            else:
                continue
        if prmp:
            prms.append(i)
    return prms

def print_primes():
    a = int(input('first number: '))
    b = int(input('second number'))
    ans = primes(a, b)
    ans = [n for n in ans if n>=a]
    print(ans)

print_primes()
````

## Excercise 4:

Write a program that, given a **list of integers**, creates a new list which contains the
initial elements of the input list until the first element which is even.

````python
def ex4(li):
    ans = []
    for elem in li:
        if elem % 2 == 1:
            ans.append(elem)
        else:
            break
    return ans

def ex4_io():
    li = []
    while True:
        ask_new_number = True
        while ask_new_number:
            try:
                n = input("Type an integer (-1 to end): ")
                n = int(n)
                ask_new_number = False
            except ValueError:
                print('Not an integer!')
        if n == -1:
            print (f"The list is {li}\nThe solution is {ex4(li)}")
            break
        else:
            li.append(n)

ex4_io()
````

## Ejercicio 5

Write a program that, given a **list of tuples**, computes a new list that contains the sums
of the elements of such tuples, and is **sorted in decreasing order.**

````python
def ex5(lt):
    ans = [sum (elem) for elem in lt]
    ans.sort(reverse=True)
    return ans
print(ex5([(8, 9), (8, -1)]))

````

## Ejercicio 6

Write a program that, given two strings, prints on the console the unique characters
that appear in the first string but not in the second string. Sets must be employed to
develop the program.

````python
def ex6(str1, str2):
    set1, set2 = set(str1), set(str2)
    return set1 - set2

print(ex6())
````

## Ejercicio 7

Write a program that, given a **dictionary and a value**, creates a new dictionary with all
the items of the input dictionary **except those that contain such value.**

````python
def ex7(dic, val):
    # Desempaquetado directo de clave (k) y valor (v)
    return {k: v for k, v in dic.items() if v != val}

d1 = {'a':1, 'b': 2, 'c': 2, 'd':3}
print(ex7(d1, 2))
````

## Ejercicio 8

Write a function to compute the **second smallest element of a list of numbers**, **without
sorting** the list entirely.

````python
import math
def ex8(ln):
    n1 = math.inf
    n2 = math.inf
    if len(ln) < 2:
        raise ValueError("List too short")
    for n in ln:
        if n <= 1:
            n1, n2 = n, n1
        elif n <= n2:
            n1, n2 = n1, n
        else:
            continue
    return n2

ln8 = [0, 1, 2, 1, 3, -1, 4, -4, 5, 7, -6, -9]
print(f"ln8 = {ln8}, \nThe second smallest element is {ex8(ln8)}")
````
