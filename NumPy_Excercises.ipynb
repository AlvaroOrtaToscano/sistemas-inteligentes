# NumPy Library Excercises

## Excercise 1:

Write a program that creates a unidimensional array with the positive integers smaller than
100 that are multiples of 4.

```python
import numpy as np

my_array1 = np.arange(0, 100, 4)
print(my_array1)
```

## Excercise 2:

Write a program that creates a bidimensional array of **5 rows and 4 columns** with the integers from **0 to 19**

```python
import numpy as np

my_array2 = np.arange(20).reshape(5, 4)
print(my_array2)
```

## Excercise 3:

Write a program that inverts the order of elements of a unidimensional array

```python
import numpy as np

my_array3 = np.arange(20)
array_inverted = np.flip(my_array3)
print(array_inverted)
```

## Excercise 4

Write a program that inverts the order of the rows of a bidimensional array

```python
import numpy as np

my_array4 = np.arange(20).reshape(5, 4)
inverted_rows = np.flipud(my_array4)
print(my_array4)
print(f"\n{inverted_rows}")
```
